### RGB to HSV

<img src="https://miro.medium.com/v2/resize:fit:1400/1*XK_UZpBH6CYHi01BnRWsFA.png" alt="HSV Python">

```python
def rgb_to_hsv_manual(arr,scale_factor):
    hsv_arr = np.zeros_like(arr, dtype=np.float32)
    for i in range(arr.shape[0]):
        for j in range(arr.shape[1]):
            r, g, b = arr[i, j] / 255.0
            c_max = max(r,g,b)
            c_min = min(r,g,b)
            delta = c_max - c_min
            
            if(delta==0):
                H = 0
            elif(c_max == r):
                    H = (60 *((g - b) / delta))
            elif(c_max == g):
                    H = 60*(2+(b-r)/delta)
            elif(c_max == b):
                    H = 60*(4+(r-g)/delta)

            if H<0:
                H = H + 360            
            if c_max == 0:
                S = 0
            else:
                S = (delta/c_max)

            V = c_max

            H = (H/2) * scale_factor
            S = (S*255) * scale_factor
            V = (V*255) * scale_factor
            
            hsv_arr[i][j] = [H,S,V]
    return np.clip(hsv_arr,0,255).astype(np.uint8)                        
```

```python
tup = plt.imread("tulips.jpg")
```

```python
tup[1][1]
```

```python
tup1 = tup.copy()
tup2 = tup.copy()
```

```python
new_image = rgb_to_hsv_manual(tup1,1)
```

### Red Rose Segmentation

```python
lr = np.array([0,120,70])
ur = np.array([15,255,255])
mask1 = cv2.inRange(new_image,lr,ur)
```

```python
type(mask1)
```

```python
mask1.shape
```

```python
lr = np.array([160,120,70])
ur = np.array([180,255,255])
mask2 = cv2.inRange(new_image,lr,ur)
```

```python
mask = mask1 | mask2
```

```python
res_img = cv2.bitwise_and(tup2,tup2,mask = mask)
```

```python
fig,axs = plt.subplots(ncols=2,nrows=1,figsize=(10,4))
axs[0].imshow(tup1) 
axs[1].imshow(res_img)
```

```python
lr = np.array([20,90,80])
ur = np.array([30,255,255])
mask3 = cv2.inRange(new_image,lr,ur)
```

```python
res_img = cv2.bitwise_and(tup1,tup1,mask = mask3)
```

```python
fig,axs = plt.subplots(ncols=2,nrows=1,figsize=(10,4))
axs[0].imshow(tup1) 
axs[1].imshow(res_img)
```
