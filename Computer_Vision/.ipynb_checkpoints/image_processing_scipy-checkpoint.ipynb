```python
from scipy import misc
from skimage import io
from scipy import ndimage
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = io.imread("images.jpeg")
```

```python
type(img)
```

```python
img.dtype
```

```python
img.shape
```

```python
img_gray = io.imread("images.jpeg",as_gray=True)
```

```python
img_gray.shape
```

```python
img[0,0]
```

```python
flippedLR = np.fliplr(img)
```

```python
flippedUD = np.flipud(img)
```

```python
plt.figure(figsize=(12, 12))
plt.subplot(1,3,1)
plt.imshow(img)
plt.subplot(1,3,2)
plt.imshow(flippedLR)
plt.subplot(1,3,3)
plt.imshow(flippedUD)
```

```python
# cmap (colormap) should only be used for grayscale or single-channel images.
plt.figure(figsize=(12, 12))
plt.subplot(1,3,1)
plt.imshow(img_gray,cmap="Greys")
plt.subplot(1,3,2)
plt.imshow(img_gray,cmap="Blues")
plt.subplot(1,3,3)
plt.imshow(img_gray,cmap="hsv")
```

```python
rotated = ndimage.rotate(img,45,reshape=False)
```

```python
plt.imshow(rotated)
```

```python
uniform_filtered = ndimage.uniform_filter(img,size=3)
```

```python
plt.imshow(uniform_filtered)
```

```python
gaussian_filtered = ndimage.gaussian_filter(img,sigma=2)
```

```python
plt.imshow(gaussian_filtered)
```

```python
median_filtered = ndimage.median_filter(img,size=4)
```

```python
plt.imshow(median_filtered)
```

```python
sobel_img = ndimage.sobel(img_gray,axis=0)
```

```python
plt.imshow(sobel_img)
```
