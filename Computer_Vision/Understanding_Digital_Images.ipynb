```python
# Importing Libraries
from skimage import io
import numpy as np
import matplotlib.pyplot as plt
from skimage import img_as_float 
```

```python
my_image = io.imread("images.jpeg")  ## Reading Images
```

```python
my_image  #uint type of image where number varies from 0 to 255
```

```python
random_image = np.random.random([500,500])  ## generating random images
```

```python
plt.imshow(random_image) # plotting random images
```

```python
float_image = img_as_float(my_image)   # Converting our uint8 into float64
```

```python
plt.imshow(float_image)
```

```python
plt.imshow(dark_image)
```
