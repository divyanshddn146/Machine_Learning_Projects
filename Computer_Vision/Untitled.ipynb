```python
from skimage import io
import matplotlib.pyplot as plt
from skimage.transform import rescale,resize,downscale_local_mean
from skimage.filters import roberts,sobel,prewitt,scharr ## Edge detecrtion filter_forward
from skimage.feature import canny
```

```python
img = io.imread("images.jpeg",as_gray=True)
```

```python
rescaled_img = rescale(img,1/4,anti_aliasing=True)
```

```python
plt.imshow(rescaled_img)
```

```python
rescaled_img.shape
```

```python
resized_img = resize(img,(100,100),anti_aliasing=True)
```

```python
plt.imshow(rescaled_img)
```

```python
downscaled_img = downscale_local_mean(img,(4,3))
```

```python
plt.imshow(downscaled_img)
```

```python
edge_roberts = roberts(img)
edge_sobel = sobel(img)
edge_prewitt = prewitt(img)
edge_schar = scharr(img)
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10,4))
axs[0][0].imshow(edge_roberts,cmap="gray")
axs[0][1].imshow(edge_sobel,cmap="gray")
axs[1][0].imshow(edge_prewitt,cmap="gray")
axs[1][1].imshow(edge_schar,cmap="gray")
```

```python
axs.shape
```

```python
edge_canny = canny(img,sigma=3)
```

```python
plt.imshow(edge_canny)
```
