```python
img = Image.open("images.jpeg")
```

```python
cropped_img = img.crop((0,0,100,100)) ## Crop image pixel by pixel
```

```python
cropped_img.save("cropped_image.jpeg")
```

```python
img90 = img.rotate(90)    ## Rotating image at 90 angle
```

```python
img90.save("image_90.jpeg")
```

```python
img45 = img.rotate(45,expand = True) ## Rotating image at 45 angle
```

```python
img45.save("image_45.jpeg")
```

```python
img_flip = img.transpose(Image.FLIP_LEFT_RIGHT)  ## Flipping the image from left to right
```

```python
img_flip.save("image_flip.jpeg")
```

```python
grey_image = img.convert("L") ## Converting image into grayscale
```

```python
grey_image.save("grey_image.jpeg")
```
